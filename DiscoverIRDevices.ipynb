```python
import broadlink
import sys
import json

# Discover devices
def discover_devices(timeout=5):
    devices = broadlink.discover(timeout=timeout)
    discovered = []
    for dev in devices:
        try:
            dev.auth()
            discovered.append({
                "name": dev.host[0],   # IP address
                "mac": dev.mac.hex(),
                "type": dev.devtype,
                "device": dev  # keep object for sending commands
            })
        except Exception as e:
            print(f"Failed to authenticate {dev.host}: {e}")
    return discovered

# Send command to a specific RM3 IP
def send_command(ip, command_hex, devices):
    for d in devices:
        if d["name"] == ip:
            dev = d["device"]
            try:
                dev.send_data(bytes.fromhex(command_hex))
                print(f"Command sent to {ip}")
                return True
            except Exception as e:
                print(f"Failed to send command: {e}")
                return False
    print(f"Device with IP {ip} not found")
    return False

if __name__ == "__main__":
    # Discover devices first
    devices = discover_devices()

    # If no arguments, just print discovered devices
    if len(sys.argv) == 1:
        output = [{"name": d["name"], "mac": d["mac"], "type": d["type"]} for d in devices]
        print(json.dumps(output))
    # If IP + command_hex provided, send the command
    elif len(sys.argv) == 3:
        target_ip = sys.argv[1]
        command_hex = sys.argv[2]
        send_command(target_ip, command_hex, devices)
    else:
        print("Usage:")
        print("  python discover_rm3.py             # Discover devices")
        print("  python discover_rm3.py <IP> <hex> # Send command")

```
